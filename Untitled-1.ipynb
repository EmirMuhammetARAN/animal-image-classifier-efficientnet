```python
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.applications import EfficientNetB0
from tensorflow.keras import layers
from tensorflow.keras.optimizers import Adam
import numpy as np
from sklearn.utils.class_weight import compute_class_weight
import gc

tf.keras.backend.clear_session()
gc.collect()

kaggle_input_path = "d:/archive (2)/raw-img"

train_ds_opt = keras.preprocessing.image_dataset_from_directory(
    kaggle_input_path,               
    image_size=(160, 160),
    batch_size=8,
    validation_split=0.2,    
    subset="training",
    seed=42
)

val_ds_opt = keras.preprocessing.image_dataset_from_directory(
    kaggle_input_path,
    image_size=(160, 160),
    batch_size=8,
    validation_split=0.2,
    subset="validation",
    seed=42
)

class_names = train_ds_opt.class_names

def calculate_class_weights(train_ds, class_names):
    all_labels = []
    for images, labels in train_ds:
        all_labels.extend(labels.numpy())
    
    class_weights = compute_class_weight(
        'balanced',
        classes=np.unique(all_labels),
        y=all_labels
    )
    
    class_weight_dict = {i: weight for i, weight in enumerate(class_weights)}
    return class_weight_dict

class_weight_dict = calculate_class_weights(train_ds_opt, class_names)

base_model = EfficientNetB0(
    weights='imagenet',
    include_top=False,
    input_shape=(160, 160, 3)
)

base_model.trainable = False

simple_model = keras.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dropout(0.2),
    layers.Dense(len(class_names), activation='softmax')
])

simple_model.compile(
    optimizer=Adam(learning_rate=0.001),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

train_ds_opt = train_ds_opt.prefetch(buffer_size=tf.data.AUTOTUNE)
val_ds_opt = val_ds_opt.prefetch(buffer_size=tf.data.AUTOTUNE)

# İlk aşama: Transfer learning (5 epoch)
print("İlk aşama: Transfer learning...")
history_simple = simple_model.fit(
    train_ds_opt,
    epochs=5,
    validation_data=val_ds_opt,
    class_weight=class_weight_dict,
    verbose=1
)

# İkinci aşama: Fine-tuning (son katmanları aç)
print("\nİkinci aşama: Fine-tuning...")
base_model.trainable = True

# Sadece son 20 katmanı eğit
for layer in base_model.layers[:-20]:
    layer.trainable = False

simple_model.compile(
    optimizer=Adam(learning_rate=0.0001),  # Daha düşük learning rate
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

history_fine = simple_model.fit(
    train_ds_opt,
    epochs=3,
    validation_data=val_ds_opt,
    class_weight=class_weight_dict,
    verbose=1
)

print(f"Final validation accuracy: {max(history_fine.history['val_accuracy']):.4f}")
```

```python
# Model kaydetme
simple_model.save('animal_classifier_model.h5')
print("Model başarıyla kaydedildi: animal_classifier_model.h5")

# Model özetini göster
simple_model.summary()

# Sınıf isimlerini kaydet
import json
with open('class_names.json', 'w') as f:
    json.dump(class_names, f)
print("Sınıf isimleri kaydedildi: class_names.json")
```

```python
def predict_file(file_path):
    img = tf.keras.preprocessing.image.load_img(file_path, target_size=(160, 160))
    img_array = tf.keras.preprocessing.image.img_to_array(img) / 255.0
    img_array = np.expand_dims(img_array, axis=0)
    pred = simple_model.predict(img_array)
    
    # Tüm tahminleri göster
    print("Tüm sınıf tahminleri:")
    for i, (class_name, prob) in enumerate(zip(class_names, pred[0])):
        print(f"{class_name}: %{prob*100:.2f}")
    
    return class_names[np.argmax(pred)], np.max(pred)

# Önce sınıfları kontrol edelim
print("Sınıf isimleri:")
print(class_names)
print(f"Toplam sınıf sayısı: {len(class_names)}")

# Model özetini tekrar göster
print("\nModel son katman çıkış boyutu:")
print(f"Dense layer units: {simple_model.layers[-1].units}")

# Test et
print("\n" + "="*50)
result, confidence = predict_file("d:/ornek.jpg")
print(f"\nEn yüksek tahmin: {result}, Güven: %{confidence*100:.2f}")

# Model ağırlıklarını kontrol et
print(f"\nModel eğitildi mi? Son katman ağırlık özeti:")
weights = simple_model.layers[-1].get_weights()[0]
print(f"Ağırlık ortalaması: {np.mean(weights):.6f}")
print(f"Ağırlık standart sapması: {np.std(weights):.6f}")
```
